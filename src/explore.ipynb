```python
import pandas as pd
import sqlite3 
import ast
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.neighbors import NearestNeighbors

# Load data into Pandas DataFrames
url1 = '/workspaces/K_Neighbours_PereHDZ/movies.csv'
url2 = '/workspaces/K_Neighbours_PereHDZ/credits.csv'

df1 = pd.read_csv(url1, sep=',')
df2 = pd.read_csv(url2, sep=',')

conn = sqlite3.connect('movies_database.db')

df1.to_sql('movies', conn, if_exists='replace', index=False)
df2.to_sql('credits', conn, if_exists='replace', index=False)

df1.to_csv('movies_data.csv', index=False)

df1.head()
```

```python
df2.to_csv('credits_data.csv', index=False)

df2.head()
```

```python
#Merge both datasets into 1
df1.to_sql('movies', conn, if_exists='replace', index=False)
df2.to_sql('credits', conn, if_exists='replace', index=False)

query = """
    CREATE TABLE IF NOT EXISTS unified_movies AS
    SELECT m.id, m.title, m.overview, m.genres, m.keywords, c.cast, c.crew
    FROM movies m
    INNER JOIN credits c ON m.title = c.title
"""

# Execute the SQL query to create the new table
cursor = conn.cursor()
cursor.execute(query)
conn.commit()

df_merged = pd.read_sql('SELECT * FROM unified_movies', conn)

# Close the database connection
conn.close()

df_merged.to_csv('cleaned_movies_data.csv', index=False)

df_merged.head()
```

```python
#Extract 'names' from JSON
def extract_names(json_str):
    try:
        data = ast.literal_eval(json_str) 
        return [item['name'].replace(' ', '') for item in data]
    except (ValueError, SyntaxError):
        return []
    

df_merged['genres'] = df_merged['genres'].apply(extract_names)
df_merged['keywords'] = df_merged['keywords'].apply(extract_names)

df_merged.head()
```

```python
#Extract cast from JSON, with no space between name and surname
def extract_cast(json_str):
    try:
        data = ast.literal_eval(json_str)
        return [item['name'].replace(' ', '') for item in data[:3]] 
    except (ValueError, SyntaxError):
        return []
    
df_merged['cast'] = df_merged['cast'].apply(extract_cast)

df_merged['cast'].head()
```

```python
#Extract the director from JSON in 'crew'
def extract_director(json_str):
    try:
        data = ast.literal_eval(json_str)
        for person in data:
            if person['job'] == 'Director':
                return person['name'].replace(' ', '')  
        return ''
    except (ValueError, SyntaxError):
        return ''
    
df_merged['crew'] = df_merged['crew'].apply(extract_director)

df_merged['crew'].head()
```

```python
# Convert overview to list of words
df_merged['overview'] = df_merged['overview'].fillna('').apply(lambda x: x.split())

df_merged['overview'].head()
```

```python
# Combine all processed columns into a single 'tags' column
df_merged["tags"] = df_merged["genres"] + df_merged["keywords"] + df_merged["cast"] + df_merged["crew"].apply(lambda x: [x]) + df_merged["overview"]

# Convert list to string, remove commas and convert to lowercase
df_merged["tags"] = df_merged["tags"].apply(lambda x: " ".join(map(str, x)).replace(",", "").lower())

# Drop the original columns
df_merged = df_merged[["id", "title", "tags"]]

df_merged.to_csv("processed_movies_data.csv", index=False)

df_merged.head()
```

```python
#Create KNN model
vectorizer = TfidfVectorizer()
matrix = vectorizer.fit_transform(df_merged["tags"])

model = NearestNeighbors(n_neighbors = 6, algorithm = "brute", metric = "cosine")
model.fit(matrix)

def get_recommendations(title):
    index = df_merged[df_merged["title"] == title].index[0]
    distances, indices = model.kneighbors(matrix[index])
    similar_movies = [(df_merged["title"][i], distances[0][j]) for j, i in enumerate(indices[0])]
    return similar_movies[1:]

#Try model
recommendations = get_recommendations('Avatar')
recommendations
```

Our model seems to work fine, since the movies it recommended are somewhat simillar to 'Avatar' (Most of the movies are about aliens, warfare or both). Just to make sure, we will try it again with a different genre:

```python
recommendations = get_recommendations('The Dark Knight Rises')
recommendations
```

Every single recommended movie is about Batman, so our model seems to work just fine.
